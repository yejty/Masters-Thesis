```python
import sympy as sp

t = sp.Symbol('t')

try:
    parametric_curve = sp.Matrix([
        sp.Piecewise((t, (0 < t) & (t < 1)), (1, (1 < t) & (t < 2))),
        sp.Piecewise((0, (0 < t) & (t < 1)), (t - 1, (1 < t) & (t < 2)))
    ])
    print(parametric_curve)
except sp.SympifyError as e:
    print(f"SympifyError: {e}")
```
